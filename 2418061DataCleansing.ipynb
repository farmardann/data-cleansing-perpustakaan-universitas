# 4.Data Cleansing-Standarisasi Kolom Nama Anggota

```python
import pandas as pd
import re

# Baca file .xlsx
file_path = '/content/drive/MyDrive/Colab Notebooks/Data-Kotor-Perpustakaan.xlsx'
df = pd.read_excel(file_path, engine='openpyxl')

# Fungsi untuk membersihkan kolom Nama Anggota
def clean_name(name):
    # Jika data kosong
    if pd.isna(name):
        return name

    # Ubah menjadi string
    name = str(name)

    # Hapus spasi di awal dan akhir
    name = name.strip()

    # Hapus spasi berlebih di tengah nama
    name = re.sub(r'\s+', ' ', name)

    # Rapikan huruf besar dan kecil
    name = name.title()

    return name

# Terapkan fungsi ke kolom Nama Anggota
df['Nama Anggota Bersih'] = df['Nama Anggota'].apply(clean_name)

# Tampilkan data sebelum dan sesudah cleansing
df[['Nama Anggota', 'Nama Anggota Bersih']].head(10)
```

# 5.Data Cleansing-Standarisasi Kolom Program Studi

```python
# Standarisasi nama program studi
def clean_prodi(prodi):
    if pd.isna(prodi):
        return prodi

    prodi = str(prodi).strip()

    prodi = re.sub(
        r'^TI$|^Informatika$|^Teknik Informatika$',
        'Teknik Informatika',
        prodi,
        flags=re.IGNORECASE
    )

    return prodi

df['Program Studi Bersih'] = df['Program Studi'].apply(clean_prodi)

df[['Program Studi', 'Program Studi Bersih']].head(10)
```

# 6.Data Cleansing-Standarisasi Kolom Status

```python
# Standarisasi Status Peminjaman Buku
def clean_status(status):
    if pd.isna(status):
        return status

    status = str(status).strip().lower()

    if status == 'dipinjam':
        return 'Dipinjam'
    elif status == 'dikembalikan':
        return 'Dikembalikan'
    else:
        return status.title()

df['Status Bersih'] = df['Status'].apply(clean_status)

df[['Status', 'Status Bersih']].head(10)
```

# 7.Data Cleansing-Standarisasi Kolom Penerbit

```python
# Standarisasi nama penerbit
def clean_penerbit(penerbit):
    if pd.isna(penerbit):
        return penerbit

    penerbit = str(penerbit).strip()

    penerbit_lower = penerbit.lower()

    if penerbit_lower in ['andi', 'andi publisher']:
        return 'Andi Publisher'
    elif penerbit_lower == 'elex media':
        return 'Elex Media Komputindo'
    else:
        return penerbit.title()

df['Penerbit Bersih'] = df['Penerbit'].apply(clean_penerbit)

df[['Penerbit', 'Penerbit Bersih']].head(10)
```

# 8.Data Cleansing-Standarisasi Kolom Tanggal Pinjam

```python
# Define a function to standardize the date format
def standardize_date(date):
    try:
        # Convert the date to datetime format
        standardized_date = pd.to_datetime(
            date,
            errors='coerce',
            dayfirst=True
        ).strftime('%d-%m-%Y')
    except:
        standardized_date = None

    return standardized_date

# Apply the standardization function to the 'Tanggal Pinjam' column
df['Tanggal Pinjam Bersih'] = df['Tanggal Pinjam'].apply(standardize_date)

# Display the DataFrame with the standardized dates
df[['Tanggal Pinjam', 'Tanggal Pinjam Bersih']].head()
```

# 11.Data Enrichment

```python
import pandas as pd

# Tahun saat ini
tahun_sekarang = 2026

# fungsi untuk mengelompokkan usia buku
def kelompok_usia_buku(tahun):
    try:
        usia = tahun_sekarang - int(tahun)

        if usia <= 2:
            return 'Buku Baru'
        elif usia <= 6:
            return 'Buku Menengah'
        elif usia <= 11:
            return 'Buku Lama'
        else:
            return 'Buku Sangat Lama'

    except:
        return 'Tidak Diketahui'

# Menerapkan fungsi untuk data enrichment
df['Kelompok Usia Buku'] = df['Tahun Terbit'].apply(kelompok_usia_buku)

# Menampilkan hasil data enrichment
df[['Judul Buku', 'Tahun Terbit', 'Kelompok Usia Buku']].head(10)
```
